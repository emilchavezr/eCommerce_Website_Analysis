### Loading Files

```python
import pandas as pd

home_df = pd.read_csv('home_page_table.csv')
search_df = pd.read_csv('search_page_table.csv')
payment_df = pd.read_csv('payment_page_table.csv')
confirmation_df = pd.read_csv('payment_confirmation_table.csv')
user_df = pd.read_csv('user_table.csv')

print("File loaded successfully.")
```

### Cleaning

```python
dataframes = [home_df, search_df, payment_df, confirmation_df, user_df]

for df in dataframes:
    df.columns = df.columns.str.strip().str.lower()
    
    if 'page' in df.columns:
        df['page'] = df['page'].str.strip().str.lower()

    df['user_id'] = df['user_id'].astype(int)

#Remove duplicates only from the user table
user_df = user_df.drop_duplicates(subset='user_id')

print("Data cleaned successfully.")
```

Preview first, before exporting them.

```python
from IPython.display import display

display(home_df.head())
display(search_df.head())
display(payment_df.head())
display(confirmation_df.head())
display(user_df.head())
```

### Exporting

```python
home_df.to_csv('clean_home.csv', index=False)
search_df.to_csv('clean_search.csv', index=False)
payment_df.to_csv('clean_payment.csv', index=False)
confirmation_df.to_csv('clean_confirmation.csv', index=False)
user_df.to_csv('clean_users.csv', index=False)

print("Clean CSVs exported.")
```
